```python
import anthropic

# Set your API key here
API_KEY = "your-api-key-here"

# We'll use Claude Haiku for speed and cost efficiency
MODEL_NAME = "claude-haiku-4-5-20251001"

# Persist these variables across notebooks
%store API_KEY
%store MODEL_NAME

print(f"API Key: {API_KEY[:10]}..." if len(API_KEY) > 10 else "Please set your API key above!")
print(f"Model: {MODEL_NAME}")
```

```python
# Test your setup with a simple API call
client = anthropic.Anthropic(api_key=API_KEY)

message = client.messages.create(
    model=MODEL_NAME,
    max_tokens=100,
    temperature=0.0,
    messages=[{"role": "user", "content": "Hello! Say 'Setup complete!' and nothing else."}]
)

print(message.content[0].text)
```

### Helper Function

Throughout this tutorial, we'll use a `get_completion` helper function. It starts simple and grows as we learn new techniques.

Each notebook defines its own version, so you can run any chapter independently.

```python
def get_completion(prompt: str):
    """Basic completion helper - sends a user message and returns Claude's response."""
    client = anthropic.Anthropic(api_key=API_KEY)
    message = client.messages.create(
        model=MODEL_NAME,
        max_tokens=2000,
        temperature=0.0,
        messages=[{"role": "user", "content": prompt}]
    )
    return message.content[0].text

# Quick test
print(get_completion("Count to three."))
```
