## Lesson

The Messages API requires three key parameters:
- **`model`**: Which Claude model to use
- **`max_tokens`**: Maximum number of tokens in the response
- **`messages`**: A list of message objects with `role` and `content`

### Rules for Messages
1. Messages must alternate between `user` and `assistant` roles
2. The first message must always have the `user` role
3. Each message needs both `role` and `content` fields

There's also an optional **system prompt** that sets Claude's behavior and personality.

```python
import anthropic

# Restore persisted variables
%store -r API_KEY
%store -r MODEL_NAME

client = anthropic.Anthropic(api_key=API_KEY)

def get_completion(prompt: str, system_prompt: str = ""):
    kwargs = {
        "model": MODEL_NAME,
        "max_tokens": 2000,
        "temperature": 0.0,
        "messages": [{"role": "user", "content": prompt}]
    }
    if system_prompt:
        kwargs["system"] = system_prompt
    message = client.messages.create(**kwargs)
    return message.content[0].text
```

### A Basic API Call

Let's start with the simplest possible interaction:

```python
# A simple prompt
response = get_completion("Hi Claude, how are you?")
print(response)
```

### Using System Prompts

System prompts let you give Claude context, personality, or behavioral instructions that apply to the entire conversation.

Without a system prompt:

```python
# Without system prompt
response = get_completion("Why is the sky blue?")
print(response)
```

```python
# With system prompt - Claude responds differently!
response = get_completion(
    "Why is the sky blue?",
    system_prompt="Respond with critical thinking questions instead of answers. Always ask follow-up questions that help the user think deeper."
)
print(response)
```

Notice how the system prompt completely changed Claude's behavior — from giving an answer to asking probing questions!

### What Happens with Bad Formatting?

Let's see some common API errors:

```python
# ERROR: Messages must start with 'user' role
try:
    message = client.messages.create(
        model=MODEL_NAME,
        max_tokens=100,
        messages=[{"role": "assistant", "content": "Hello"}]
    )
except Exception as e:
    print(f"Error: {e}")
```

```python
# ERROR: Roles must alternate
try:
    message = client.messages.create(
        model=MODEL_NAME,
        max_tokens=100,
        messages=[
            {"role": "user", "content": "Hello"},
            {"role": "user", "content": "How are you?"}
        ]
    )
except Exception as e:
    print(f"Error: {e}")
```
