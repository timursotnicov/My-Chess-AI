# Chapter 2: Being Clear and Direct

## Lesson

The **Golden Rule of Clear Prompting**: Show your prompt to a colleague — if they'd be confused about what to do, Claude will be too.

Claude has no context beyond what you provide. It can't read your mind or make assumptions about your intent. The more specific and direct you are, the better Claude performs.

### Key Techniques
- **Be specific**: Instead of "write something about dogs", say "write a 4-line poem about golden retrievers playing in the park"
- **Eliminate ambiguity**: Give Claude enough context to understand exactly what you want
- **Direct instructions**: You can simply tell Claude to skip preambles, be concise, or give a definitive answer

```python
import anthropic

%store -r API_KEY
%store -r MODEL_NAME

client = anthropic.Anthropic(api_key=API_KEY)

def get_completion(prompt: str, system_prompt: str = ""):
    kwargs = {
        "model": MODEL_NAME,
        "max_tokens": 2000,
        "temperature": 0.0,
        "messages": [{"role": "user", "content": prompt}]
    }
    if system_prompt:
        kwargs["system"] = system_prompt
    message = client.messages.create(**kwargs)
    return message.content[0].text
```

### Example: Removing Preambles

Claude often starts responses with introductory text. You can eliminate this by simply asking:

```python
# Without direction - Claude may add a preamble
response = get_completion("Write a poem about the ocean.")
print("--- With preamble ---")
print(response)
print()

# With direction - skip the preamble
response = get_completion("Write a poem about the ocean. Skip the preamble; go straight into the poem.")
print("--- Without preamble ---")
print(response)
```

### Example: Forcing a Definitive Answer

Claude sometimes hedges when asked subjective questions. You can push for a direct answer:

```python
# Vague question - Claude may hedge
response = get_completion("Who is the best basketball player of all time?")
print("--- Hedging response ---")
print(response)
print()

# Direct question - force a pick
response = get_completion(
    "Who is the best basketball player of all time? "
    "Yes, there are differing opinions, but if you absolutely had to pick one player, who would it be?"
)
print("--- Direct response ---")
print(response)
```
