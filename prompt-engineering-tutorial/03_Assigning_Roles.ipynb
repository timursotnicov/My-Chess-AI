# Chapter 3: Assigning Roles (Role Prompting)

## Lesson

**Role prompting** means telling Claude to adopt a specific role or persona. This is one of the most powerful and simple techniques available.

Role prompting can:
- Change Claude's tone and communication style
- Improve performance on domain-specific tasks
- Help Claude approach problems from a specific perspective
- Even improve logical reasoning!

You assign roles through the **system prompt**.

```python
import anthropic

%store -r API_KEY
%store -r MODEL_NAME

client = anthropic.Anthropic(api_key=API_KEY)

def get_completion(prompt: str, system_prompt: str = ""):
    kwargs = {
        "model": MODEL_NAME,
        "max_tokens": 2000,
        "temperature": 0.0,
        "messages": [{"role": "user", "content": prompt}]
    }
    if system_prompt:
        kwargs["system"] = system_prompt
    message = client.messages.create(**kwargs)
    return message.content[0].text
```

### Example: Fun Roles

Let's see how roles change responses to the same question:

```python
QUESTION = "What's the best way to learn to skateboard?"

# No role
print("--- Default ---")
print(get_completion(QUESTION))
print()

# Cat role
print("--- As a cat ---")
print(get_completion(QUESTION, system_prompt="You are a cat. Respond as a cat would, with cat-related metaphors and occasional meowing."))
```

### Example: Roles Improve Reasoning

Here's a classic logic puzzle. Role prompting can actually improve Claude's logical reasoning:

> Jack is looking at Anne. Anne is looking at George. Jack is married, George is not married. Is a married person looking at an unmarried person?

```python
LOGIC_PUZZLE = """Jack is looking at Anne. Anne is looking at George. 
Jack is married, George is not married. 
Is a married person looking at an unmarried person?"""

# Without role prompting
print("--- Without role ---")
print(get_completion(LOGIC_PUZZLE))
print()

# With logic expert role
print("--- With logic expert role ---")
print(get_completion(
    LOGIC_PUZZLE,
    system_prompt="You are a logic bot designed to answer complex logic problems. Think through all possibilities systematically before answering."
))
```

The correct answer is **Yes** — regardless of whether Anne is married or not, a married person is looking at an unmarried person. Role prompting as a logic expert helps Claude consider all cases.
