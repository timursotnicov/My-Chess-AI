# Chapter 5: Formatting Output & Speaking for Claude

## Lesson

Two powerful techniques in this chapter:

### 1. Structured Output
You can ask Claude to output in specific formats: XML, JSON, markdown tables, CSV, etc.

### 2. Prefilling (Speaking for Claude)
You can put text in the **assistant turn** to guide Claude's response. This is called "prefilling" — you're essentially starting Claude's response for it.

Common uses:
- Force a specific output format (start with `{` for JSON)
- Continue a pattern (start with an XML tag)
- Steer Claude's perspective or approach

```python
import anthropic

%store -r API_KEY
%store -r MODEL_NAME

client = anthropic.Anthropic(api_key=API_KEY)

def get_completion(prompt: str, system_prompt: str = "", prefill: str = ""):
    """Extended helper with prefill support."""
    messages = [{"role": "user", "content": prompt}]
    if prefill:
        messages.append({"role": "assistant", "content": prefill})
    kwargs = {
        "model": MODEL_NAME,
        "max_tokens": 2000,
        "temperature": 0.0,
        "messages": messages
    }
    if system_prompt:
        kwargs["system"] = system_prompt
    message = client.messages.create(**kwargs)
    return message.content[0].text
```

### Example: Requesting XML Output

```python
# Ask Claude to put a haiku in XML tags
response = get_completion("Write a haiku about robots. Put it in <haiku> tags.")
print(response)
```

### Example: Prefilling to Skip Preamble

```python
# Prefill with the opening tag — Claude continues directly
response = get_completion(
    "Write a haiku about robots. Put it in <haiku> tags.",
    prefill="<haiku>"
)
print("<haiku>" + response)
```

### Example: Forcing JSON Output

```python
# Without prefill
print("--- Without prefill ---")
response = get_completion("Give me a JSON object with keys 'name', 'age', and 'city' for a fictional person.")
print(response)
print()

# With prefill — forces JSON format
print("--- With JSON prefill ---")
response = get_completion(
    "Give me a JSON object with keys 'name', 'age', and 'city' for a fictional person.",
    prefill="{"
)
print("{" + response)
```

### Example: Combining Techniques

Let's combine XML tags for input, output formatting, and prefilling:

```python
EMAIL = "Hey, just wanted to follow up on the project deadline. Can we push it back a week? Let me know ASAP!"

response = get_completion(
    f"Rewrite this email in Old English style. <email>{EMAIL}</email>\n\nPut your response in <olde_english_email> tags.",
    prefill="<olde_english_email>"
)
print("<olde_english_email>" + response)
```

### Exercise 5.2
Modify the `PROMPT` to produce **two** haikus, each wrapped in its own `<haiku>` tags.

```python
# Exercise 5.2
PROMPT = "[Your prompt here]"

response = get_completion(PROMPT)
print(response)

# Grading
def grade_exercise_5_2(response):
    return response.count("<haiku>") >= 2 and len(response.split("\n")) > 5

print("\n" + ("✅ PASS" if grade_exercise_5_2(response) else "❌ TRY AGAIN — Need two haikus in <haiku> tags"))
```

### Exercise 5.3
Write a prompt template that produces two haikus about two **different** animals using `{ANIMAL1}` and `{ANIMAL2}` variables.

```python
# Exercise 5.3
ANIMAL1 = "cat"
ANIMAL2 = "dog"

PROMPT = f"[Your prompt template here using {ANIMAL1} and {ANIMAL2}]"

response = get_completion(PROMPT)
print(response)

# Grading
def grade_exercise_5_3(response, animal1, animal2):
    resp_lower = response.lower()
    return (animal1 in resp_lower or animal2 in resp_lower) and response.count("<haiku>") >= 2

print("\n" + ("✅ PASS" if grade_exercise_5_3(response, ANIMAL1, ANIMAL2) else "❌ TRY AGAIN"))
```
